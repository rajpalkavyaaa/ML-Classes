## Question 09
_The age column has missing values. Using only shape and info(), work out what percentage of age values are missing. Write the number in a comment_

```python
total_rows = t.shape[0]
non_missing_age = 714  # From t.info()

missing_age = total_rows - non_missing_age
missing_percentage = (missing_age / total_rows) * 100

print(missing_age)
print(missing_percentage)

# Age has 177 missing values, approximately 19.87%.
```
