```python
# Ques-->09 Award medals to the scores below using nested np.where: 'Gold' for 90 and above, 'Silver' for 75 and above, 'Bronze' for 60 and above, and 'None' otherwise. Then count how many students received each medal.

scores = np.array([94, 61, 77, 45, 90, 88, 59, 72])
medals = np.where(scores>=90,"Gold",
         np.where(scores>=75,"Silver",
                  np.where(scores>=60,"Bronze","None")))

print("Gold:", np.sum(medals == "Gold"))
print("Silver:", np.sum(medals == "Silver"))
print("Bronze:", np.sum(medals == "Bronze"))
print("None:", np.sum(medals == "None"))
```

```python
# Ques-->14 Build a complete multi-step pipeline. Starting from the two arrays below: convert them into column vectors, join them into a single two-column table, attach a third column holding each row's total, then multiply the three-column table by w = [0.5, 0.2, 0.3] to obtain one score per row. Finally use np.where to mark each score as 1 if it is above the average score and 0 otherwise. Report the shape of every intermediate array.

hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])

hours_col = hours.reshape(-1, 1)
attendance_col = attendance.reshape(-1, 1)

X = np.hstack((hours_col,attendance_col))

totals = X.sum(axis=1)
totals = totals.reshape(-1, 1)

X_3 = np.hstack((X, totals))

w = [0.5, 0.2, 0.3]
scores = X_3 @ w

avg = scores.mean()

result = np.where(scores > avg, 1, 0)
print(result)
```

```python
# Ques-->17 28 daily temperature readings are given. Reshape them into 4 weeks of 7 days, then: find each week's average, identify the hottest week, and count how many individual days crossed 35 degrees.

np.random.seed(7)
temps = np.random.randint(15, 41, 28)

temps = np.reshape(temps,(4,7))
avg = np.mean(temps,axis=1)
hottest = np.argmax(temps)
hot_days = np.sum(temps > 35)

print(avg)
print(hottest)
print(hot_days)
```
