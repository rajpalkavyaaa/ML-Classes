## 01. What is Pandas?

_Pandas is a Python library used to work with tabular data—data arranged in rows and columns, like an Excel sheet._


```python
# Pandas puts names on the rows and columns
arr = np.array([
    [16.99, 1.01],
    [10.34, 1.66],
    [21.01, 3.50]
])
table = pd.DataFrame(
    arr,
    columns=["total_bill", "tip"]
)
print(table.loc[1, "tip"])
print(table.loc[2,"total_bill"])
```

## 02. Series — one labelled column
_A Series is one-dimensional data with labels attached._

```python
sales = pd.Series([4200, 5100, 8900, 7600])
print(sales)

# We can provide meaningful index labels:
sales = pd.Series(
    [4200, 5100, 8900, 7600],
    index=["Thur", "Fri", "Sat", "Sun"]
)
print(sales["Sat"])
print(sales.values)
print(sales.dtype)
print(sales.shape)
```

## 03. DataFrame — a complete table
_A DataFrame is two-dimensional data containing rows and columns._

```python
data = {
    "name": ["Kavya", "Riya", "Aman"],
    "marks": [85, 72, 91],
    "city": ["Pune", "Mumbai", "Delhi"]
}

df = pd.DataFrame(data)
print(df.shape)           #DataFrame.shape = (rows, columns)
print(df["marks"])        #Complete table → DataFrame
                          #One column → Series
print(type(df["marks"])) #Although df is a DataFrame, one selected column is a Series
print()
```

## 04. Creating a DataFrame from a dictionary
- Each dictionary key becomes a column name.
- Each list becomes the data stored in that column.
- Values at matching list positions form one row.

```python
data = {
    "table": [1, 2, 3, 4],
    "amount": [450, 1200, 300, 890],
    "waiter": ["Ravi", "Sunita", "Ravi", "Amit"]
}

orders = pd.DataFrame(data)

# We can also replace the default 0, 1, 2, 3 index:
orders = pd.DataFrame(
    data,
    index=["a", "b", "c", "d"]
)
print(orders.loc["b", "amount"])

# This will produce an error:
data = {
    "table": [1, 2, 3, 4],       # 4 values
    "amount": [450, 1200, 300]   # 3 values
}
orders.loc["c", "waiter"] # ERROR!
```

## 07. How do head() and tail() inspect a DataFrame?
- After loading a dataset, we should inspect some records before analysing it.
- head() — view the beginning
- tail() — view the end

```python
# The tips.csv dataset has 244 rows, with default indexes starting from 0
# First index = 0
# Last index  = 243


df = pd.read_csv("tips.csv")

print(df.head()) # Returns the first 5 rows by default.
print(df.head(3)) # Returns the first 3 rows.
print(df.tail()) # Returns the last 5 rows by default.
print(df.tail(2)) # Returns the last 2 rows
```

## 08. How do we check the size, column names, and data types?


```python
df = pd.read_csv("tips.csv")
print(df.shape)

print(df.shape[0])   # 244 rows
print(df.shape[1])      # 7 columns
print(df.columns)      # To see column names
print(list(df.columns)) # To get them as a regular Python list




# df.shape   ---------->  # correct
# df.shape() ---------->  # incorrect


# shape is an attribute, so we do not use parentheses.
# Attribute → stores/provides information: df.shape, df.columns, df.dtypes
# Method → performs an action: df.head(), df.tail()
```

## 09. What does info() tell us about a DataFrame?

_info() gives us a quick health check of the dataset. It shows:_

- Number of rows
- Column names
- Number of non-null values

- nMemory usage

```python
# RangeIndex: 244 entries, 0 to 243
# Data columns: 7 columns

# Column       Non-Null Count    Dtype
# total_bill   244 non-null      float64
# tip          244 non-null      float64
# sex          244 non-null      object
# size         244 non-null      int64


# Non-null means the value is present—it is not missing.
# If there are 244 rows and a column shows
# 238 non-null
# Then its missing values are:
# 244 - 238 = 6

df = pd.read_csv("tips.csv")
print(df.info())

# df.info() ---------> # correct
# df.info   ---------> # does not execute the method
```

## 10. What does describe() tell us about a DataFrame?
_describe() gives a statistical summary of every numeric column_

```python
# | Result  | Meaning                      |
# | ------- | ---------------------------- |
# | `count` | Number of non-missing values |
# | `mean`  | Average                      |
# | `std`   | Standard deviation           |
# | `min`   | Smallest value               |
# | `25%`   | First quartile               |
# | `50%`   | Median                       |
# | `75%`   | Third quartile               |
# | `max`   | Largest value                |


# For the tips dataset, the numeric columns are:
# total_bill
# tip
# size
df = pd.read_csv("tips.csv")

# Text columns such as sex, day, and time do not appear in the default result because calculating their average is impossible.

df.describe()                   #→ numeric columns
df.describe(include="object")   #→ text columns
df.describe(include="all")      #→ all columns
```

## 11. What does value_counts() do?
_value_counts() tells us how many times each unique value appears in one column._

```python
df = pd.read_csv("tips.csv")
df["day"].value_counts()
df["day"].value_counts(normalize=True) #This gives proportions, such as 0.356, instead of counts such as 87, which is 86/244
```

## 12. What is the difference between unique(), nunique(), and value_counts()?
- unique() — Which different values exist?
- nunique() — How many different values exist?
- value_counts() — How many times does each value appear?

```python
df = pd.read_csv("tips.csv")
print(df["day"].unique())
print(df["day"].nunique())
print(df["day"].value_counts())


# Command	Question answered
# unique()	Which values?
# nunique()	How many different values?
# value_counts()	How many times does each value occur?
```
